# Step 3 — Deploy Web Interface (Colab)

Launches the Flask backend and exposes it via [localtunnel](https://theboroer.github.io/localtunnel-www/) so you can test the web UI from any browser.

**Prerequisites:**
- `model.keras` — trained model from `2_train.ipynb`
- `static/index.html` — the web UI (included in this repo)
- `app.py` — the Flask backend (included in this repo)

```python
import os, shutil, subprocess, threading, time

# Make sure static/ folder exists and index.html is inside it
os.makedirs("static", exist_ok=True)
if os.path.exists("index.html") and not os.path.exists("static/index.html"):
    shutil.move("index.html", "static/index.html")

print("model.keras found:", os.path.exists("model.keras"))
print("static/index.html found:", os.path.exists("static/index.html"))
```

```python
# ── Start Flask in a background thread ───────────────────────────────────────
def run_flask():
    subprocess.run(["python", "app.py"])

t = threading.Thread(target=run_flask, daemon=True)
t.start()
time.sleep(8)   # wait for model to load

# ── Expose via localtunnel ────────────────────────────────────────────────────
lt  = subprocess.Popen(["lt", "--port", "5000"], stdout=subprocess.PIPE)
url = lt.stdout.readline().decode().strip()
print("\nApp URL:", url)
```
